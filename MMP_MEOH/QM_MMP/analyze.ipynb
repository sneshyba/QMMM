```python
import pickle
import os
import sys; sys.path.append('/home/chemistry/QMMM/')
import visualize_QMMM as vQ
import SCF_helpers as Sh
import py3Dmol
```

```python
with open("MMP_QM.pkl", "rb") as f:
    qm_data = pickle.load(f)

mol_opt = qm_data["mol_opt"]
mo_coeff = qm_data["mo_coeff"]
mo_energy = qm_data["mo_energy"]
mo_occ = qm_data["mo_occ"]
n_occ = qm_data["n_occ"]

orbital_filenames = qm_data["orbital_filenames"]
```

```python
viewwidth = 500
viewheight = 500

print("n_occ =", n_occ)

for iorbital in range(n_occ + 1):

    orbital_filename = orbital_filenames[iorbital]

    view = py3Dmol.view(
        width=viewwidth,
        height=viewheight
    )

    vQ.show_orbital(
        view,
        orbital_filename,
        iorbital,
        mol_opt
    )

    vQ.show_QM_atoms(
        view,
        mol_opt
    )

    view.show()
```
