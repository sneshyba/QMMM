```python
import numpy as np
from pyscf import gto, scf, qmmm
```

```python
# ============================================================
# Constants
# ============================================================

HARTREE_TO_KJMOL = 2625.49962
```

```python
# ============================================================
# 1. QM SYSTEM
#
# methanol
#
# ============================================================

# This has to be changed to become the coordinates of methanol

mol = gto.M(
    atom='''
        O    0.000    0.000    0.000
        H    0.970    0.000    0.000
    ''',
    basis='sto-3g',
    charge=-1,
    spin=0,
    unit='Angstrom',
    verbose=4
)
```

```python
# ============================================================
# 3. BARE QM CALCULATION
# ============================================================

mf_qm = scf.RHF(mol)

energy_qm_hartree = mf_qm.kernel()

energy_qm_kjmol = (
    energy_qm_hartree * HARTREE_TO_KJMOL
)
```

```python
# ============================================================
# 9. PRINT RESULTS
# ============================================================

print()
print("Bare QM energy:")
print(
    f"    {energy_qm_hartree:.12f} Hartree"
)
print(
    f"    {energy_qm_kjmol:.6f} kJ/mol"
)

```
